```python
import os
import pandas as pd
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_experimental.agents import create_pandas_dataframe_agent
```

```python
# 1. Setup API Keys (ensure these are in your Colab Secrets 🔑)
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
```

```python
import kagglehub
path = kagglehub.dataset_download("toxifarian/tesco-retail-store-clubcard-orders")
```

```python
print(path)
```

```python
df = pd.read_csv(path+'/Tesco.csv')
```

```python
df.head()
```

```python
# 3. Initialize the LLM (Using 2026 models like GPT-4o or GPT-5 mini)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
# 4. Create the Agent
# IMPORTANT: 'allow_dangerous_code=True' is required in 2026 to execute Python analysis
agent = create_pandas_dataframe_agent(
    llm,
    df,
    verbose=True,
    allow_dangerous_code=True
)
```

```python
# 5. Query the Data
response = agent.invoke("give me sample 5 questions i can ask about this dataset?")
print(response['output'])
```

```python
# 5. Query the Data
response = agent.invoke("How many customers have left the membership, and what percentage does that represent of the total customer base?")
print(response['output'])
```
