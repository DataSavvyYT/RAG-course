```python
# 3. Load the EmbeddingGemma model
# This 300M parameter model is free and optimized for semantic similarity
model_id = "google/embeddinggemma-300m"
model = SentenceTransformer(model_id)
```

```python
# The two sentences with "interest" in different contexts
sentence1 = "I have no interest in football."
sentence2 = "Bank interest rate is 4%."
```

```python
# 5. Generate embeddings
# The model encodes the entire sentence, capturing the contextual meaning of "interest"
embed1 = model.encode(sentence1)
embed2 = model.encode(sentence2)
```

```python
# 2. Get embeddings for the word "interest" in isolation (Static-like baseline)
embed_word_only = model.encode("interest")
```

```python
# 6. Calculate Cosine Similarity (1.0 = identical, 0.0 = completely different)
sim_between_sentences = 1 - cosine(embed1, embed2)
sim_word_to_s1 = 1 - cosine(embed_word_only, embed1)
sim_word_to_s2 = 1 - cosine(embed_word_only, embed2)
```

```python
print(f"--- Results for {model_id} ---")
print(f"Similarity between Sentence 1 and 2: {sim_between_sentences:.4f}")
print(f"Similarity ('interest' alone vs Sentence 1): {sim_word_to_s1:.4f}")
print(f"Similarity ('interest' alone vs Sentence 2): {sim_word_to_s2:.4f}")
```

```python
# Analysis
if sim_between_sentences < 0.7:
    print("\nObservation: The low similarity between the two sentences confirms that the model")
    print("treats the context of 'interest' (hobby vs. finance) as distinct concepts.")
```
